```python
jacksondata = '/home/dean/Downloads/OMEandSingleCellMasks/OMEnMasks/ome'
```

```python
import tifffile as tif
```

```python
import os

#list all files in the directory
files = os.listdir(jacksondata)
```

```python
#get the mean image size and the number of channels

image_sizes = []
for i in files:
    image = tif.imread(jacksondata + '/' + i)
    image_sizes.append(image.shape)

print(image_sizes)
```

```python
#compute the mean image dimension
import numpy as np
mean_image_dimension = np.mean(image_sizes, axis=0)
print(mean_image_dimension)

```
